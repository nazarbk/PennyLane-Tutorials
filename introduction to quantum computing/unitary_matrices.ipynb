```python
import pennylane as qp
import numpy as np
```

```python
# Codercise l.3.1 - Unitaries in PennyLane

dev = qp.device("default.qubit", wires=1)

U = np.array([[1, 1], [1, -1]]) / np.sqrt(2)


@qp.qnode(dev)
def apply_u():

    qp.QubitUnitary(U, wires=0)

    # Return the state
    return qp.state()

apply_u()
```

```python
# Codercise l.3.2 - Parametrized unitaries

dev = qp.device("default.qubit", wires=1)


@qp.qnode(dev)
def apply_u_as_rot(phi, theta, omega):

    qp.Rot(phi, theta, omega, wires=0)

    return qp.state()
```
