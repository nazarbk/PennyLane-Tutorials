```python
import pennylane as qp
import numpy as np
```

```python
dev = qp.device("default.qubit", wires=2)

# Starting from the state |00>, implement a PennyLane circuit
# to construct each of the Bell basis states.


@qp.qnode(dev)
def prepare_psi_plus():
    ##################
    # YOUR CODE HERE #
    ##################

    qp.H(0)
    qp.CNOT(wires=[0, 1])

    return qp.state()


@qp.qnode(dev)
def prepare_psi_minus():
    ##################
    # YOUR CODE HERE #
    ##################

    qp.X(0)
    qp.H(0)
    qp.CNOT(wires=[0, 1])

    return qp.state()


@qp.qnode(dev)
def prepare_phi_plus():
    ##################
    # YOUR CODE HERE #
    ##################

    qp.X(1)
    qp.H(0)
    qp.CNOT(wires=[0, 1])
    

    return qp.state()


@qp.qnode(dev)
def prepare_phi_minus():
    ##################
    # YOUR CODE HERE #
    ##################

    qp.X(0)
    qp.X(1)
    qp.H(0)
    qp.CNOT(wires=[0, 1])

    return qp.state()


psi_plus = prepare_psi_plus()
psi_minus = prepare_psi_minus()
phi_plus = prepare_phi_plus()
phi_minus = prepare_phi_minus()

# Uncomment to print results
print(f"|ψ_+> = {psi_plus}")
print(f"|ψ_-> = {psi_minus}")
print(f"|ϕ_+> = {phi_plus}")
print(f"|ϕ_-> = {phi_minus}")
```

```python
# Codercise l.14.2 - Quantum Multiplexer

dev = qp.device("default.qubit", wires=3)

# State of first 2 qubits
state = [0, 1]


@qp.qnode(device=dev)
def apply_control_sequence(state):
    # Set up initial state of the first two qubits
    if state[0] == 1:
        qp.PauliX(wires=0)
    if state[1] == 1:
        qp.PauliX(wires=1)

    # Set up initial state of the third qubit - use |->
    # so we can see the effect on the output
    qp.PauliX(wires=2)
    qp.Hadamard(wires=2)

    ##################
    # YOUR CODE HERE #
    ##################

    # IMPLEMENT THE MULTIPLEXER
    qp.MultiControlledX(wires=[0, 1, 2], control_values=[0, 1])

    qp.H(2)
    qp.MultiControlledX(wires=[0, 1, 2], control_values=[1, 0])
    qp.H(2)

    qp.ctrl(qp.Y(2), control=[0, 1])

    return qp.state()


print(apply_control_sequence(state))
```
