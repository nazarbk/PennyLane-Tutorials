```python
import pennylane as qp
import numpy as np
```

```python
# Pauli gates
qp.PauliX(wires=0)

qp.PauliY(wires=0)

qp.PauliZ(wires=0)

# Hadamard Gate
qp.Hadamard(wires=0)
```

```python
# Example 1
dev = qp.device("default.qubit", wires=1)

@qp.qnode(dev)
def my_circuit():
    qp.Hadamard(wires=0)
    qp.PauliZ(wires=0)
    return qp.state()

print(qp.draw(my_circuit)())
```

```python
# Example 2
dev = qp.device("default.qubit", wires=1)

@qp.qnode(dev)
def flip_circuit():
    qp.X(wires=0)
    return qp.state()

print(qp.draw(flip_circuit)())
print(flip_circuit())
```

```python
# Rotation gates
angle = (np.pi / 2)

qp.RY(angle, wires=0)

qp.RX(angle, wires=0)

qp.RZ(angle, wires=0)
```

```python
# Example 3
dev = qp.device("default.qubit", wires=1)

@qp.qnode(dev)
def parametized_circuit(theta):
    qp.RX(theta, wires=0)
    return qp.expval(qp.PauliZ(0))

angle = np.pi / 2
result = parametized_circuit(angle)

print(result)
```
