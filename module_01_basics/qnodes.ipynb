```python
import pennylane as qp
```

```python
# 1. We define the device (a 1-qubit simulator)
dev = qp.device("default.qubit", wires=1)

# 2. We create the QNode using the decorator and passing it the device
@qp.qnode(dev)
def my_quantum_circuit(param):
    qp.RX(param, wires=0)
    return qp.expval(qp.PauliZ(0))
```

```python
print(my_quantum_circuit(0.5))
```

```python
# Example 1
dev = qp.device("default.qubit", wires=1)

@qp.qnode(dev)
def my_quantum_circuit(theta):
    qp.RY(theta, wires=0)
    return qp.expval(qp.PauliX(0))

print(my_quantum_circuit(0.5))
```
