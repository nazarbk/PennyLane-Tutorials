```python
import pennylane as qp
import numpy as np
```

```python
dev = qp.device("default.qubit", wires=2)

def my_first_quantum_function(theta):
    qp.RX(theta, wires = 0)
    qp.PauliY(wires = 1)
    qp.Hadamard(wires = 0)
    qp.Hadamard(wires = 1)

    return qp.state()

my_first_qnode = qp.QNode(my_first_quantum_function, dev)
```

```python
print(my_first_qnode(np.pi/4))
```

```python
# Other ways to specify the number of wires
dev = qp.device("default.qubit", wires=3)

dev = qp.device("default.qubit", wires=[0, 1, 2])

dev = qp.device("default.qubit", wires=range(3))

# We can change the order of the wires
dev_scrambled = qp.device("default.qubit", wires=[2, 0, 1])

# We also can name the wires
dev_c_t = qp.device("default.qubit", wires=["target", "control"])
```

```python
@qp.qnode(dev_c_t)
def create_entangled():
    qp.Hadamard(wires="control")
    qp.CNOT(wires=["control", "target"])

    return qp.state()
```

```python
# We can build large circuits with functions
def subcircuit1(angle):
    qp.RX(angle, wires=0)
    qp.PauliY(wires=1)

def subcircuit2():
    qp.Hadamard(wires=0)
    qp.CNOT(wires=[0, 1])

def full_circuit(theta, phi):
    subcircuit1(theta)
    subcircuit2()
    subcircuit1(phi)
```

```python
# We can plot it
theta = 0.3
phi = 0.2

print(qp.draw(full_circuit)(theta, phi))
```

```python
# Codercise PF.1.1a - Writing a quantum function

def circuit(angle):
    qp.Hadamard(wires=0)
    qp.PauliX(wires=1)
    qp.CNOT(wires=[0, 1])
    qp.RY(angle, wires=1)

    return qp.state()

print(qp.draw(circuit)(angle=0.2))
```

```python
# Codercise PF.1.1b - Creating devices

dev_qubit = qp.device("default.qubit", wires=["alice", "bob"])
dev_mixed = qp.device("default.mixed", wires=2)

@qp.qnode(dev_qubit)
def example_circuit(theta):
    
    qp.RX(theta, wires =  "alice")
    qp.CNOT(wires =  ["alice", "bob"])
    
    return qp.state()

print(example_circuit(0.3))
```

```python
# Codercise PF.1.1c - Circuit into QNode

dev = qp.device("default.qubit", wires=2)

circuit_qnode = qp.QNode(circuit, dev)

print(circuit_qnode(0.3))
```

```python
# Codercise PF.1.2a - Defining subcircuits

def subcircuit_1(angle):
    qp.RX(angle, wires=0)
    qp.PauliY(wires=1)

def subcircuit_2():
    qp.Hadamard(wires=0)
    qp.CNOT(wires=[0, 1])

dev = qp.device('default.qubit', wires = 2)

@qp.qnode(dev)
def full_circuit(theta, phi):
    subcircuit_1(theta)
    subcircuit_2()
    subcircuit_1(phi)

    return qp.state()
```

```python
# Codercise PF.1.2b - Jumbled wires

def subcircuit_1(angle, wire_list):
    """
    Implements the first subcircuit as a function of the RX gate angle
    and the list of wires wire_list on which the gates are applied
    """

    qp.RX(angle, wires=wire_list[0])
    qp.PauliY(wires=wire_list[1])

def subcircuit_2(wire_list):
    """
    Implements the second subcircuit as a function of the list of wires 
    wire_list on which the gates are applied
    """

    qp.Hadamard(wires=wire_list[0])
    qp.CNOT(wires=wire_list)

dev = qp.device("default.qubit", wires = [0,1])

@qp.qnode(dev)
def full_circuit(theta, phi):
    """
    Builds the full quantum circuit given the input parameters
    """

    subcircuit_1(theta, [0, 1])
    subcircuit_2([0, 1])
    subcircuit_1(phi, [1, 0])

    return qp.state()
```
