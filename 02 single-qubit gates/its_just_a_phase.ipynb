```python
import pennylane as qp
import numpy as np
```

```python
dev = qp.device("default.qubit", wires=1)


@qp.qnode(dev)
def apply_z_to_plus():
    """Write a circuit that applies PauliZ to the |+> state and returns
    the state.

    Returns:
        np.array[complex]: The state of the qubit after the operations.
    """

    ##################
    # YOUR CODE HERE #
    ##################

    # CREATE THE |+> STATE
    qp.Hadamard(wires=0)

    # APPLY PAULI Z
    qp.Z(wires=0)

    # RETURN THE STATE
    return qp.state()


print(apply_z_to_plus())
```

```python
dev = qp.device("default.qubit", wires=1)


@qp.qnode(dev)
def fake_z():
    """Use RZ to produce the same action as Pauli Z on the |+> state.

    Returns:
        np.array[complex]: The state of the qubit after the operations.
    """

    ##################
    # YOUR CODE HERE #
    ##################

    qp.Hadamard(wires=0)

    # APPLY RZ
    qp.RZ(np.pi, wires=0)

    # RETURN THE STATE
    return qp.state()

print(fake_z())
```

```python
dev = qp.device("default.qubit", wires=1)


@qp.qnode(dev)
def many_rotations():
    """Implement the circuit depicted above and return the quantum state.

    Returns:
        np.array[complex]: The state of the qubit after the operations.
    """

    ##################
    # YOUR CODE HERE #
    ##################

    qp.Hadamard(wires=0)
    qp.S(wires=0)
    qp.adjoint(qp.T)(wires=0)
    qp.RZ(0.3, wires=0)
    qp.adjoint(qp.S)(wires=0)

    return qp.state()

print(many_rotations())
```
