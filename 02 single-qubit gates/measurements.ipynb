```python
# Codercise l.9.2 - Y basis rotation

# WRITE A QUANTUM FUNCTION THAT PREPARES (1/2)|0> + i(sqrt(3)/2)|1>
def prepare_psi():
    # or qp.RX(-2 * np.pi / 3, wires=0)
    state = np.array([ 1 / 2, (np.sqrt(3) / 2) * 1.0j])
    qp.MottonenStatePreparation(state, wires=0)


# WRITE A QUANTUM FUNCTION THAT SENDS BOTH |0> TO |y_+> and |1> TO |y_->
def y_basis_rotation():
   qp.H(wires=0)
   qp.S(wires=0) 
```

```python
# Codercise l.9.3 - Measurement in the Y basis

dev = qp.device("default.qubit", wires=1)

@qp.qnode(dev)
def measure_in_y_basis():
    ##################
    # YOUR CODE HERE #
    ##################

    prepare_psi()

    qp.adjoint(y_basis_rotation)()

    return qp.probs(wires=0)


print(measure_in_y_basis())
```
